### Categorical Data Validation

The Education and Marital_Status columns were inspected for inconsistent values and formatting.

Unnecessary whitespace was removed from text fields. Existing category labels were preserved because the dataset alone does not provide sufficient evidence to determine whether unusual category labels should be merged or renamed.

```python
text_columns = df.select_dtypes(include=["object"]).columns

for col in text_columns:
    df[col] = df[col].str.strip()
```

```python
print("Missing Income before cleaning:",
      df["income"].isnull().sum())
```

```python
income_median = df["income"].median()

print("Income median:", income_median)
```

```python
df["income"] = df["income"].fillna(income_median)
```

```python
print("Missing Income after cleaning:",
      df["income"].isnull().sum())
```

```python
df["dt_customer"].head()
```

```python
df["dt_customer"] = pd.to_datetime(
    df["dt_customer"],
    dayfirst=True,
    errors="coerce"
)
```

```python
print(df["dt_customer"].dtype)
```

```python
print(
    "Invalid dates:",
    df["dt_customer"].isnull().sum()
)
```

```python
numeric_columns = [
    "year_birth",
    "income",
    "kidhome",
    "teenhome",
    "recency",
    "mntwines",
    "mntfruits",
    "mntmeatproducts",
    "mntfishproducts",
    "mntsweetproducts",
    "mntgoldprods"
]
```

```python
for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")
```

```python
print("Negative income:",
      (df["income"] < 0).sum())

print("Negative recency:",
      (df["recency"] < 0).sum())

print("Negative wine spending:",
      (df["mntwines"] < 0).sum())

print("Negative fruit spending:",
      (df["mntfruits"] < 0).sum())
```

```python
# Check for potential Income outliers using the IQR method

Q1 = df["income"].quantile(0.25)
Q3 = df["income"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

income_outliers = df[
    (df["income"] < lower_bound) |
    (df["income"] > upper_bound)
]

print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)
print("Potential Income outliers:", len(income_outliers))

income_outliers[["id", "income"]].head(10)
```

### Outlier Check

Potential Income outliers were identified using the Interquartile Range (IQR) method. These observations were not automatically removed because an outlier can represent a genuine customer value rather than a data-entry error. Further business context would be required before removing or modifying such records.

```python
print("Minimum birth year:", df["year_birth"].min())
print("Maximum birth year:", df["year_birth"].max())
```

```python
# Check for potentially unusual birth years

current_reference_year = 2014

age_check = current_reference_year - df["year_birth"]

print("Minimum calculated age:", age_check.min())
print("Maximum calculated age:", age_check.max())

print("\nCustomers with calculated age above 100:")
df.loc[age_check > 100, ["id", "year_birth"]]
```

### Birth Year Data Quality Check

The `Year_Birth` column was inspected for potentially unusual values.

Some very old birth years were identified in the dataset. These records were not automatically removed because an unusual value does not necessarily mean that it is incorrect. The records were retained because there was insufficient evidence within the dataset to confirm that they were data-entry errors.

## Data Cleaning Summary

The raw Customer Personality Analysis dataset initially contained 2,240 rows and 29 columns.

The following preprocessing steps were performed:

1. Inspected the dataset structure, dimensions, and data types.
2. Identified missing values.
3. Found 24 missing values in the `Income` column.
4. Replaced missing `Income` values using the median income value of 51,381.5.
5. Checked for duplicate rows; no duplicate rows were found.
6. Checked for duplicate customer IDs; no duplicate IDs were found.
7. Standardized column names using lowercase letters and underscores.
8. Removed unnecessary whitespace from text fields.
9. Inspected categorical values for consistency.
10. Converted `Dt_Customer` to the `datetime64[ns]` data type.
11. Checked for invalid dates.
12. Validated important numerical columns.
13. Checked for negative values in relevant numerical fields.
14. Investigated potentially unusual birth-year values.
15. Performed final missing-value and duplicate checks.

The final dataset contains 2,240 rows and 29 columns and has no missing values or duplicate rows.
